```python
import pandas as pd
 
filepath = r"data_analysis\results_2025-08-09_20_episodes\results_for_run_dated_2025-08-09_14-51_1.csv"

df = pd.read_csv(filepath)

list_of_seqs = df['Original Sequence'].tolist()

def convert_sequence_list_to_fasta(sequence_list):
    fasta_output_string = ""
    for index, sequence_data in enumerate(sequence_list):
        fasta_output_string += f">sequence_{index + 1}\n{sequence_data}\n"
    return fasta_output_string

fasta_output_string = convert_sequence_list_to_fasta(list_of_seqs)

output_filepath = "test_sequences.fasta"

with open(output_filepath, "w") as file_object:
    file_object.write(fasta_output_string)

```
